```python
# Check placeholder values in Cast dataset
placeholder_values = ["Unknown", "unknown", "N/A", "NA", "None", "null", ""]

for column in ["actor_name", "character_name"]:
    print(f"\n{column}:")
    print(cast[column][cast[column].isin(placeholder_values)].value_counts())
```

```python
# Replace Unknown with NaN in character_name
cast["character_name"] = cast["character_name"].replace(
    ["Unknown", "unknown"], pd.NA
)

# Verify missing values
print("Missing Character Names:",
      cast["character_name"].isnull().sum())
```

```python
print("Duplicate Rows:", cast.duplicated().sum())
```

```python
# Check duplicate cast_row_id values
print("Duplicate Cast IDs:", cast["cast_row_id"].duplicated().sum())
```

```python
# Find orphan cast records
orphan_cast = cast[~cast["movie_id"].isin(movies["movie_id"])]

# Count orphan records
print("Orphan Cast Records:", len(orphan_cast))

# Remove orphan records if any exist
cast = cast[cast["movie_id"].isin(movies["movie_id"])].copy()

# Verify after removal
print("Cast Dataset Shape After Cleaning:", cast.shape)
```

```python
import os
import pandas as pd
from sqlalchemy import create_engine, inspect, text

# Create the data folder if it doesn't exist yet
os.makedirs("data", exist_ok=True)

engine = create_engine("sqlite:///data/TMDB_database.db")
```

```python
# Store dates as plain YYYY-MM-DD and money columns as integers
movies["release_date"] = pd.to_datetime(movies["release_date"]).dt.strftime("%Y-%m-%d")
movies["budget"] = movies["budget"].astype("int64")
movies["revenue"] = movies["revenue"].astype("int64")

# Load all six tables into the database
movies.to_sql("movies", engine, if_exists="replace", index=False)
genres.to_sql("genres", engine, if_exists="replace", index=False)
movie_genres.to_sql("movie_genres", engine, if_exists="replace", index=False)
cast.to_sql("cast", engine, if_exists="replace", index=False)
crew.to_sql("crew", engine, if_exists="replace", index=False)
movie_keywords.to_sql("movie_keywords", engine, if_exists="replace", index=False)
print("All six tables loaded.")
```

```python
inspector = inspect(engine)
tables = inspector.get_table_names()

print("Available Tables:", tables)
print("Total Tables:", len(tables))
```
